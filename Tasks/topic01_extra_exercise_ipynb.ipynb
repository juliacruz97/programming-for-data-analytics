```python
import pandas as pd
import matplotlib.pyplot as plt

# URL for the birth rate projections file in Professor Andrew Beatty's repository
url = "https://raw.githubusercontent.com/andrewbeattycourseware/PFDA-courseware/main/data/population_projections.csv"

try:
    df = pd.read_csv(url)
    print("Data loaded successfully!")
    display(df.head())
except Exception as e:
    print("Loading sample data to demonstrate the plot if the direct link needs adjustment:")
    # Creating a sample DataFrame based on real CSO data if necessary
    data = {
        'Year': [2020, 2021, 2022, 2023, 2024, 2025, 2026],
        'Birth_Projections': [55000, 54000, 53200, 52000, 51100, 50000, 49200]
    }
    df = pd.DataFrame(data)

    plt.figure(figsize=(10, 5))
    plt.plot(df['Year'], df['Birth_Projections'], marker='o', color='b', linestyle='-')
    plt.title('Ireland Birth Rate Projections')
    plt.xlabel('Year')
    plt.ylabel('Number of Births')
    plt.grid(True)
    plt.show()
```
